```python
#OPERATORS

a = 17
b = 5

print(a + b)    # 22  — Addition
print(a - b)    # 12  — Subtraction
print(a * b)    # 85  — Multiplication
print(a / b)    # 3.4 — Division (ALWAYS float)
print(a // b)   # 3   — Floor Division (removes decimal, rounds DOWN)
print(a % b)    # 2   — Modulo (remainder)
print(a ** b)   # 1419857 — Exponentiation (17 to the power 5)
```

```python
print(17 // 5)     # 3   — because 17/5 = 3.4, floor = 3
print(-17 // 5)    # -4  — because -17/5 = -3.4, floor = -4 (towards negative infinity!)
print(17 // -5)    # -4  — same logic
```

```python
print(17 % 5)    # 2  — 17 = 5×3 + 2
print(10 % 2)    # 0  — even numbers give 0
print(10 % 3)    # 1  — 10 = 3×3 + 1
```

```python
# Check if number is even or odd
num = 17
if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
print(2 ** 10)    # 1024
print(4 ** 0.5)   # 2.0 — square root! (because x^0.5 = √x)
print(8 ** (1/3)) # 2.0 — cube root!
```

```python
a = 10
b = 20

print(a == b)    # False — equal to
print(a != b)    # True  — not equal to
print(a > b)     # False — greater than
print(a < b)     # True  — less than
print(a >= b)    # False — greater than or equal to
print(a <= b)    # True  — less than or equal to
```

```python
x = 15
print(10 < x < 20)     # True  — Python allows this! (Java/C don't)
print(10 < x < 14)     # False
```

```python
a = True
b = False

print(a and b)    # False — both must be True
print(a or b)     # True  — at least one must be True
print(not a)      # False — flips the value
```

```python
# 'and' returns the first falsy value, or the last value if all are truthy
print(5 and 10)       # 10   — 5 is truthy, so returns 10
print(0 and 10)       # 0    — 0 is falsy, returns it immediately
print("" and "hi")    # ""   — empty string is falsy

# 'or' returns the first truthy value, or the last value if all are falsy
print(5 or 10)        # 5    — 5 is truthy, returns it immediately
print(0 or 10)        # 10   — 0 is falsy, moves to 10
print(0 or "")        # ""   — both falsy, returns last value
```

```python
# These all evaluate to False in a boolean context:
False
0
0.0
""          # empty string
[]          # empty list
{}          # empty dict
()          # empty tuple
None
```

```python
x = 10         # basic assignment

x += 5         # same as x = x + 5   → x is now 15
x -= 3         # same as x = x - 3   → x is now 12
x *= 2         # same as x = x * 2   → x is now 24
x /= 4         # same as x = x / 4   → x is now 6.0
x //= 2        # same as x = x // 2  → x is now 3.0
x **= 3        # same as x = x ** 3  → x is now 27.0
x %= 5         # same as x = x % 5   → x is now 2.0
```

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)    # True  — values are same
print(a is b)    # False — they are different objects in memory!
print(a is c)    # True  — c points to the SAME object as a
```

```python
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)       # True
print("grape" in fruits)       # False
print("grape" not in fruits)   # True

name = "Divya"
print("iv" in name)            # True — works on strings too!
```
